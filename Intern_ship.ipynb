```python
import sqlite3

conn = sqlite3.connect("scout.db")
cursor = conn.cursor()
print(conn)
```

```python
cursor.execute("""
    CREATE TABLE jobs (
        id INTEGER PRIMARY KEY,
        title TEXT,
        company TEXT,
        location TEXT,
        job_type TEXT,
        salary TEXT,
        requirements TEXT,
        url TEXT UNIQUE
    );
""")
conn.commit()
```

```python
import json

sql = "INSERT INTO jobs (title, company, location, job_type, salary, requirements, url) VALUES (?, ?, ?, ?, ?, ?, ?)"

for d in all_jobs:
    values = (
        d["title"],
        d["company"],
        d["location"],
        d["job-type"],
        d["salary"],
        json.dumps(d["Requirement"]),
        d["url"]
    )
    cursor.execute(sql, values)

conn.commit()
print("Inserted", len(all_jobs), "jobs")
```

```python
# 1. Rebuild job fresh
job = []
url = "https://www.jobberman.com/jobs?experience=graduate-trainee"
for page_num in range(1, 6):
    listing = f"{url}&page={page_num}"
    resp = requests.get(listing)
    soup = BeautifulSoup(resp.text, "html.parser")
    for i in soup.find_all("a", href=True):
        link = i.get('href')
        if "/listings/" in link:
            job.append(link)
print("job unique:", len(set(job)))
```

```python
all_jobs = []
for l in job:
    pest = requests.get(l)
    ment = pest.text
    reel = BeautifulSoup(ment, 'html.parser')
    job_title = reel.find('h1', {"data-cy": "title-job"})
    company = reel.find('h2')
    det = reel.find('div', class_="flex flex-wrap gap-x-6 gap-y-4 mt-4")
    tags = det.find_all('a', href=True)
    deta = reel.find('span', class_="inline-flex items-center gap-x-2 text-base leading-6 font-normal text-gray-900")
    meta = reel.find_all('span', class_=["text-sm leading-5 font-semibold text-gray-800 md:text-base md:leading-6", "text-sm leading-5 font-normal text-gray-800 md:text-base md:leading-6 ml-1"])
    labels = meta[0::2]
    values = meta[1::2]
    requirements = {}
    for lab, val in zip(labels, values):
        requirements[lab.text.strip()] = val.text.strip()
    one_job = {"title": job_title.text.strip(), "company": company.text.strip(), "location": tags[0].text.strip(), "job-type": tags[1].text.strip(), "salary": deta.text.strip(), "Requirement": requirements, "url": l}
    all_jobs.append(one_job)

urls_check = [d["url"] for d in all_jobs]
print("all_jobs total:", len(all_jobs))
print("all_jobs unique urls:", len(set(urls_check)))
```
